### Vector Databases, Embeddings, RAG
1. Embedding - An embedding is a numerical representation of the semantic meaning of data.
2. Vector Database - A vector database stores embeddings and allows us to search for vectors that are semantically similar.
3. Document - A document is a piece of info stored in pdfs, webpages, databases etc, in langchain its an object 

```python
from langchain_core.documents import Document

documents = Document(
    page_content="Passengers can cancel their booking before departure.",
    metadata={
        "source": "cancellation_policy.pdf",
        "page": 12
    }
)
documents
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

text = "I want to book a flight"

vector = embeddings.embed_query(text)

print(len(vector))
print(vector[:10])
```

```python
embeddings.embed_documents([
    "Jet2 allows flight cancellations.",
    "Passengers can purchase baggage."
])
```

### Creating a Chroma Vector Store

```python
from langchain_core.documents import Document
from langchain_chroma import Chroma

documents = [
    Document(
        page_content="Jet2 allows passengers to change their flight."
    ),
    Document(
        page_content="Passengers can purchase additional baggage."
    ),
    Document(
        page_content="Flight cancellations can be requested through customer support."
    ),
]

vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    collection_name="flight_docs",
)
```

```python
results = vectorstore.similarity_search(
    "How can I change my flight?",
    k=2
)

for doc in results:
    print(doc.page_content)
```

### Retriever

A Retriever is responsible for retrieving relevant documents based on a query.

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 2}
)
results = retriever.invoke(
    "How can I change my flight?"
)
```
