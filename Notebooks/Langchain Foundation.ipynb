### Runnables
1. A Runnable is essentially a component that: Takes an input → does something → produces an output
2. chain = prompt | model | parser, is a runnable as the output from left goes to right

```python
## This literally does no change to the input, good for passing questions to the prompt
from langchain_core.runnables import RunnablePassthrough

chain = RunnablePassthrough()

result = chain.invoke(
    "Hello"
)

print(result)
```

```python
### Run multiple branches using the same input.

from langchain_core.runnables import RunnableParallel

chain = RunnableParallel(
    first=lambda x: x + 1,
    second=lambda x: x * 2,
)

result = chain.invoke(10)

print(result)
```

```python
## 
from langchain_core.runnables import RunnableParallel,RunnablePassthrough

chain = RunnableParallel(
    question = RunnablePassthrough(), ## Simply passing x will error out
    length = lambda x:len(x) 
    )

chain.invoke('what is rag')
```

```python
from langchain_core.runnables import RunnableLambda

def uppercase(text):
    return text.upper()

uppercase_runnable = RunnableLambda(uppercase)
result = uppercase_runnable.invoke("hello")
print(result)
```

```python
from langchain_core.runnables import RunnableBranch

technical = lambda x : f"Technical answer for {x}"
general = lambda x: f"General answer for {x}"

branch = RunnableBranch(
    (
        lambda x: x['type'] == 'technical',
        technical,
    ),
    general,
)

branch.invoke({
    'type':'technical',
    'question': 'What is an api'
    })
    
```

| Runnable              | Purpose                        |
| --------------------- | ------------------------------ |
| `RunnableSequence`    | A → B → C                      |
| `RunnableParallel`    | Same input → multiple branches |
| `RunnablePassthrough` | Pass input unchanged           |
| `RunnableLambda`      | Put Python logic into a chain  |
| `RunnableBranch`      | Conditional routing            |


```python
# Sequence: # A → B → C

              #        ┌→ A
# Parallel:   # Input ─┼→ B
              #        └→ C

# Passthrough: Input → SAME INPUT

# Lambda:  Input → Python function → Output

              #            ┌→ A
# Branch:     # Input → IF ┤
              #            └→ B
```

```python
chain = RunnableParallel(
    question = RunnablePassthrough(), ## Simply passing x will error out
    length = lambda x:len(x),
    upper_case = lambda x: x.upper()
    )

chain.invoke('what is rag')
```

### PydanticOutputParser

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel


class FlightRecommendation(BaseModel):
    flight_number: str
    recommended_fare: float
    confidence: float
    direction: str
    reasons: list[str]


llm = ChatOllama(
    model="gemma4:e2b",
    temperature=0,
)

parser = PydanticOutputParser(
    pydantic_object=FlightRecommendation
)

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are a flight pricing expert."
        ),
        (
            "human",
            """
Analyze this flight:

{flight}

{format_instructions}
"""
        ),
    ]
)

prompt = prompt.partial(
    format_instructions=parser.get_format_instructions()
)

chain = prompt | llm | parser

result = chain.invoke(
    {
        "flight": """
        Flight: LS123
        Current fare: 135
        Load factor: 0.82
        DTD: 5
        ROS: 1.25
        """
    }
)

print(result)
print("-"*20)
print(result.flight_number)
print(result.recommended_fare)
print(result.confidence)
print(result.direction)
print(result.reasons)
```

### Structured Output

```python
## In structured ouptut the output format is designed without additional parsing step
## However if we want to apply any conditions like >0 or value in 1, 2 , 3 then this wont be useful

from pydantic import BaseModel
from langchain_ollama import ChatOllama


class FlightRecommendation(BaseModel):
    flight_number: str
    recommended_fare: float
    confidence: float
    direction: str
    reasons: list[str]


llm = ChatOllama(
    model="gemma4:e2b",
    temperature=0,
)

structured_llm = llm.with_structured_output(
    FlightRecommendation
)

result = structured_llm.invoke(
    """
    Analyze flight LS123.

    Current fare: £135
    Load factor: 82%
    DTD: 5
    ROS: 1.25

    Recommend the optimal fare.
    """
)

# print(result)
print("-"*40)
print(result.flight_number)
print(result.recommended_fare)
print(result.confidence)
print(result.direction)
print(result.reasons)
```

```python
from typing import Literal
from pydantic import BaseModel, Field


class FlightRecommendation(BaseModel):

    flight_number: str

    recommended_fare: float = Field(
        gt=0,
        description="Recommended fare in GBP"
    )

    confidence: float = Field(
        ge=0,
        le=1,
        description="Confidence between 0 and 1"
    )

    direction: Literal[
        "increase",
        "decrease",
        "hold",
    ]

    reasons: list[str] = Field(
        min_length=1
    )


llm = ChatOllama(model="gemma4:e2b", temperature=0)

parser = PydanticOutputParser(pydantic_object=FlightRecommendation)

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are a flight pricing expert."
        ),
        (
            "human",
            """
Analyze this flight:

{flight}

{format_instructions}
"""
        ),
    ]
)

prompt = prompt.partial(format_instructions=parser.get_format_instructions())

chain = prompt | llm | parser

result = chain.invoke(
    {
        "flight": """
        Flight: LS123
        Current fare: 135
        Load factor: 0.82
        DTD: 5
        ROS: 1.25
        """
    }
)

print(result)
print("-"*20)
print(result.flight_number)
print(result.recommended_fare)
print(result.confidence)
print(result.direction)
print(result.reasons)
```
